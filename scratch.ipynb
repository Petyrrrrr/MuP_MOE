```python
import torch
#checks if cuda is available
print(torch.cuda.is_available())
#print gpu name
print(torch.cuda.get_device_name(0))
#print gpu number
print(torch.cuda.current_device())
#print gpu memory
print(torch.cuda.memory_summary(device=None, abbreviated=False))
#print gpu memory used
print(torch.cuda.memory_allocated(device=None))
#print gpu memory reserved
print(torch.cuda.memory_reserved(device=None))
#print gpu memory cached
print(torch.cuda.memory_cached(device=None))
```
